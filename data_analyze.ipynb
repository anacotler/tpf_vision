```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv(r'utkface_dataset.csv')	
age = data['age']
plt.hist(age, bins=116, histtype='bar', rwidth=0.8)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution')
plt.show()
```

```python
ethnicity = data['ethnicity']
plt.hist(ethnicity, bins=116, histtype='bar', rwidth=0.8)
plt.xlabel('Ethnicity')
plt.ylabel('Frequency')
plt.title('Ethnicity Distribution')
plt.show()
```
